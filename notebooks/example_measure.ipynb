# Measure object volumes in the example data

The pattern for project analysis: parameters first, the work done by functions
from the package, and a figure at the end. It uses the synthetic example volume
(two objects with 8 × 8 × 16 nm voxels), so you can check the answer by hand:
object 1 is 3,276,800 nm³ and object 2 is 19,660,800 nm³.

```python
# Parameters: change these for your own data.
from pathlib import Path

DATA_DIR = Path("../data")
LABELS = DATA_DIR / "example_labels.ome.zarr"
LEVEL = 0  # pyramid level: 0 is full resolution
```

```python
# Make the example volume if it isn't there yet.
import runpy

if not LABELS.exists():
    runpy.run_path("../examples/make_example_data.py")["main"](DATA_DIR)
```

```python
from vem_project.io import read_ome_zarr
from vem_project.measure import object_volumes

labels = read_ome_zarr(LABELS, level=LEVEL)  # lazy: nothing is loaded yet
print(f"Shape {labels.data.shape}, voxel size {labels.voxel_size} {labels.unit}")

volumes = object_volumes(labels)
volumes
```

```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(4, 3))
ax.bar([str(label) for label in volumes], [v / 1e6 for v in volumes.values()])
ax.set_xlabel("Object label")
ax.set_ylabel(f"Volume (10⁶ {labels.unit}³)")
ax.set_title("Object volumes")
fig.tight_layout()
```
